```python
import pandas as pd
import numpy as np
import os
```

```python
PHENO_CSV = "/gpfs01/home/shxyt1/data/sticklebacks/pheno/qtl_all_phenotype_data.csv"
SEX_CSV = "/gpfs01/home/shxyt1/data/sticklebacks/pheno/Sex_calling_summary.csv"
GENO_FAM = "/gpfs01/home/shxyt1/data/sticklebacks/PLINK/stickleback_allchr.fam"

OUTDIR = "/gpfs01/home/shxyt1/data/sticklebacks/pheno"
TRAITDIR = f"{OUTDIR}/traits"

os.makedirs(OUTDIR, exist_ok=True)
os.makedirs(TRAITDIR, exist_ok=True)

print("All folders ready.")
```

```python
fam = pd.read_csv(GENO_FAM, sep=r"\s+", header=None)
geno_ids = set(fam[1])
len(geno_ids)
```

```python
df = pd.read_csv(PHENO_CSV)

# rename fish_id → IID
df.rename(columns={"fish_id": "IID"}, inplace=True)
df["FID"] = df["IID"]

# 仅保留 genotype 样本
df_clean = df[df["IID"].isin(geno_ids)].copy()

# 整理列顺序
cols = ["FID", "IID"] + [c for c in df_clean.columns if c not in ("FID", "IID")]
df_clean = df_clean[cols]

# 输出
clean_path = f"{OUTDIR}/all_pheno_clean.txt"
df_clean.to_csv(clean_path, sep="\t", index=False)

print(f"[OK] Clean phenotype file saved: {clean_path}")
print(f"Rows: {df_clean.shape[0]}  Columns: {df_clean.shape[1]}")
```

```python
exclude = {
    "FID","IID",
    "date_fish_caught","date_fish_processed",
    "sex","notes","icpms_run","schisto"
}

traits = [c for c in df_clean.columns if c not in exclude]

print(f"Found {len(traits)} traits.")

for t in traits:
    sub = df_clean[["FID", "IID", t]]
    outpath = f"{TRAITDIR}/pheno_{t}.txt"
    sub.to_csv(outpath, sep="\t", index=False, header=False)
    print(" ✔", outpath)
```

```python
sex_df = pd.read_csv(SEX_CSV)

# 映射性别编码
sex_df["SEX"] = sex_df["sex"].map({"M": 1, "F": 2})

# 仅保留 genotype 样本
sex_clean = sex_df[sex_df["individual"].isin(geno_ids)].copy()

# PLINK 格式
sex_clean["FID"] = sex_clean["individual"]
sex_clean["IID"] = sex_clean["individual"]

sex_out = sex_clean[["FID", "IID", "SEX"]]
sex_out_path = f"{OUTDIR}/sex.txt"
sex_out.to_csv(sex_out_path, sep="\t", index=False, header=False)

print(f"[OK] Sex file saved: {sex_out_path}")
print(sex_out.head())
```

```python
PCA_FILE  = "/gpfs01/home/shxyt1/data/sticklebacks/PLINK/stickleback_all_clean.eigenvec"

pca = pd.read_csv(PCA_FILE, sep=r"\s+", header=None)

# 构建列名
cols = ["FID", "IID"] + [f"PC{i}" for i in range(1, pca.shape[1]-1)]
pca.columns = cols

# 仅取前 5 个 PC
pc5 = pca[["FID", "IID", "PC1", "PC2", "PC3", "PC4", "PC5"]]

pc5_path = f"{OUTDIR}/pc_covariates.txt"
pc5.to_csv(pc5_path, sep="\t", index=False)

print(f"[OK] PC1–PC5 covariate file saved: {pc5_path}")
pc5.head()
```
