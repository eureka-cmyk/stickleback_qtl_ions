```python
import pandas as pd
```

```python
fam_path = "/gpfs01/home/shxyt1/data/sticklebacks/PLINK/stickleback_all_clean.fam"
pheno_path = "/gpfs01/home/shxyt1/data/sticklebacks/pheno/traits/pheno_Na_RINT.txt"
```

```python
fam = pd.read_csv(fam_path, sep=" ", header=None)
pheno = pd.read_csv(pheno_path, sep="\t", header=None, names=["FID","IID","pheno"])
```

```python
fam.columns = ["FID","IID","father","mother","sex","pheno_placeholder"]
```

```python
# merge
merged = fam.merge(pheno, on=["FID","IID"], how="left")
```

```python
# phenotype must be in column 6
fam_new = merged.iloc[:, :5].copy()
fam_new[5] = merged["pheno"]
```

```python
out_fam = "/gpfs01/home/shxyt1/data/sticklebacks/gemma/Na_RINT/stickleback_all_clean_Na_RINT.fam"
fam_new.to_csv(out_fam, sep=" ", header=False, index=False)

```
